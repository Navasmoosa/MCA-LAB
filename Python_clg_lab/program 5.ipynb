```python
n=int(input("Enter a number: "))
for i in range(1,n+1):
    for j in range(1,i+1):
        print(j*i,end=" ")
    print()
```

```python
for i in range(1,6):
    for j in range(1,i+1):
        print("*",end=" ")
    print()
    for i in range(5,1):
        for j in range(6,i-1):
            print("*",end=" ")
        print()
```

```python
square=lambda a:a*a
rectangle=lambda l,b:l*b
triangle=lambda b,h:0.5*b*h

while True:
    
    print("1.Square: ")
    print("2.Rectangle: ")
    print("3.Triangle: ")
    
    ch=int(input("Enter a choice: "))
    if(ch==1):
        a=int(input("Enter a number: "))
        print("area of the square: ",square(a))
    elif(ch==2):
        l=int(input("Enter a length: "))
        b=int(input("Enter a breadth: "))
        print("area of the Rectangle: ",rectangle(l,b))
    elif(ch==3):
        b=int(input("Enter a breadth: "))
        h=int(input("Enter a height: "))
        print("area of the Rectangle: ",triangle(b,h))
```
