```python
year=int(input("Enter year: "))
for i in range(2026,year):
    if(i%400==0 or (i%4==0 and i%100!=0)):
        print(i,"is a leap year")

```

```python
list=[1,-2,4,9,-3]
positive=[i for i in list if(i>0)]
print("positive numbers are: ",positive)

square=[i**2 for i in list]
print("Squre is: ",square)

str=input("Enter a word: ")
vovels=['A','a','E','e','I','i','O','o','U','u']
x=[i for i in str if i in vovels]
print("The vovels is: ",x)

word="hello"
orgnl_val=[ord(i) for i in word]
print("The orginal values are: ",orgnl_val)
```
